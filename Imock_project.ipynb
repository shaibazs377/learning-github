# Outlier Treatment (NOT just detection)

```python
for col in num_cols:
    Q1 = a[col].quantile(0.25)
    Q3 = a[col].quantile(0.75)
    IQR = Q3 - Q1
    
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    a[col] = a[col].clip(lower, upper)
```
